```python
import cv2
from ultralytics import YOLO
import os

# Load the YOLOv8 pose estimation model
model = YOLO('yolov8n-pose.pt')  # Use the pose-specific model

def process_image(image_path, output_dir):
    """Process a single image for pose estimation."""
    image = cv2.imread(image_path)
    results = model(image)
    annotated_image = results[0].plot()

    # Save the processed image
    output_path = os.path.join(output_dir, os.path.basename(image_path))
    cv2.imwrite(output_path, annotated_image)
    print(f"Pose estimation saved for {image_path} at {output_path}")

def process_video(video_path, output_path):
    """Process a video for pose estimation."""
    cap = cv2.VideoCapture(video_path)
    frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = int(cap.get(cv2.CAP_PROP_FPS))

    # Video writer for output
    out = cv2.VideoWriter(output_path, cv2.VideoWriter_fourcc(*'mp4v'), fps, (frame_width, frame_height))

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        # Perform pose estimation
        results = model(frame)
        annotated_frame = results[0].plot()
        out.write(annotated_frame)

    cap.release()
    out.release()
    print(f"Pose estimation video saved at {output_path}")

def main(input_path):
    """Main function to process images or videos."""
    output_dir = "output"
    os.makedirs(output_dir, exist_ok=True)

    # Check if the input is a video or image
    if input_path.endswith(('.mp4', '.avi', '.mov')):
        output_video_path = os.path.join(output_dir, "output_video.mp4")
        process_video(input_path, output_video_path)
    elif input_path.endswith(('.jpg', '.jpeg', '.png')):
        process_image(input_path, output_dir)
    else:
        print("Unsupported file type. Please provide a video or image file.")

# Example Usage
if __name__ == "__main__":
    input_path = "/content/data/run.mov" # Replace with your input file path (image or video)
    main(input_path)
```
